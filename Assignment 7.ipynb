Make a class called Employee. The __init__() method for Employee should store its different attributes: name, number, address, salary, post, year of joining, education, and his year of experience.

Make a method called describe_employee() that prints the detail Information about employee, a method update_ year_of_experience(), and a method called increase_salary() that do increamnet in it's salary.

Make an instance called employee101 from your class. Print the four attri-
butes individually, and then call all it's methods for describing employee, ssalary increamnet and increase year of experience.

Create 10 different instances from the class, and call describe_employee() for each
instance. Then do some updations with it's methods.



```python

class Employee:
    def __init__(self, name, number, address, salary, post,
                 year_of_joining, education, year_of_experience):

        self.name = name
        self.number = number
        self.address = address
        self.salary = salary
        self.post = post
        self.year_of_joining = year_of_joining
        self.education = education
        self.year_of_experience = year_of_experience

    def describe_employee(self):
        print("------ Employee Details ------")
        print("Name:", self.name)
        print("Employee Number:", self.number)
        print("Address:", self.address)
        print("Salary:", self.salary)
        print("Post:", self.post)
        print("Year of Joining:", self.year_of_joining)
        print("Education:", self.education)
        print("Year of Experience:", self.year_of_experience)

    def update_year_of_experience(self, years):
        self.year_of_experience += years
        print("Experience updated successfully!")

    def increase_salary(self, amount):
        self.salary += amount
        print("Salary increased successfully!")


employee101 = Employee(
    "Ali Ahmed", 101, "Karachi",
    50000, "Software Developer",
    2022, "BS Computer Science", 3
)

print("Name:", employee101.name)
print("Number:", employee101.number)
print("Address:", employee101.address)
print("Salary:", employee101.salary)

employee101.describe_employee()

employee101.increase_salary(10000)
employee101.update_year_of_experience(1)

employee101.describe_employee()


employee1 = Employee("Shahbaz", 102, "Karachi", 45000, "Accountant", 2021, "B.Com", 4)

employee2 = Employee("Fatima", 103, "Lahore", 60000, "HR Manager", 2020, "MBA", 5)

employee3 = Employee("Asghar", 104, "Islamabad", 55000, "Web Developer", 2022, "BS IT", 3)

employee4 = Employee("Ayesha", 105, "Karachi", 70000, "Project Manager", 2019, "MBA", 6)

employee5 = Employee("Bilal", 106, "Multan", 40000, "Sales Executive", 2023, "BBA", 2)

employee6 = Employee("Hina", 107, "Lahore", 65000, "Graphic Designer", 2021, "BS Design", 4)

employee7 = Employee("Hamza", 108, "Peshawar", 50000, "Data Analyst", 2022, "BS Statistics", 3)

employee8 = Employee("Muskan", 109, "Quetta", 75000, "Team Leader", 2018, "MS Computer Science", 7)

employee9 = Employee("Zain", 110, "Hyderabad", 35000, "Office Assistant", 2024, "Intermediate", 1)

employee10 = Employee("Maryam", 111, "Faisalabad", 58000, "Software Engineer", 2021, "BS Computer Science", 4)


employees = [
    employee1, employee2, employee3, employee4, employee5,
    employee6, employee7, employee8, employee9, employee10
]

for employee in employees:
    employee.describe_employee()


employee1.increase_salary(5000)
employee2.update_year_of_experience(1)
employee3.increase_salary(10000)
employee4.update_year_of_experience(2)
employee5.increase_salary(3000)

print("------ Updated Employee Records ------")

employee1.describe_employee()
employee2.describe_employee()
employee3.describe_employee()
employee4.describe_employee()
employee5.describe_employee()
```
