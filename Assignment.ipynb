```python

questions = [{
    "question": "How satisfied are you with our product or service?",
    "rating": 0
},
{
    "question": "How often do you use our product or service?",
    "rating": 0
},
{
    "question": "What do you like most about our product or service?",
    "rating": 0
},
{
    "question": "What problems or difficulties have you experienced while using it?",
    "rating": 0
},
{
    "question": "What improvements or new features would you like to see in the future?",
    "rating": 0
}]

for question in questions:
    print(question["question"])

    while True:
        selected_rating = input(
            "Select Rating:\n"
            "1: Poor\n"
            "2: Good\n"
            "3: Very Good\n"
            "4: Excellent\n"
        )

        try:
            selected_rating = int(selected_rating)

            if selected_rating < 1 or selected_rating > 4:
                print("Error: Please enter a rating between 1 and 4.")
            else:
                question["rating"] = selected_rating
                break

        except ValueError:
            print("Error: Invalid input! Please enter a numeric value.")

    print("Your Rating:", question["rating"])
    print("------------------------")

print("Final Survey Results:")

total_rating = 0

for question in questions:
    total_rating += question["rating"]

average_rating = total_rating / len(questions)

print("\nAverage Rating:", average_rating)

if average_rating > 4:
    print("Excellent Experience")
elif average_rating >= 3:
    print("Good Experience")
elif average_rating >= 2:
    print("Average Experience")
else:
    print("Need Improvement")
```
