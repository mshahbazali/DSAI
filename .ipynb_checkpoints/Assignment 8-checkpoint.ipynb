```python
products = {
    "P1": {"name": "Shoes", "price": 2500, "stock": 5},
    "P2": {"name": "Shirt", "price": 1500, "stock": 10},
    "P3": {"name": "Bag", "price": 3000, "stock": 4},
    "P4": {"name": "Watch", "price": 4000, "stock": 3},
    "P5": {"name": "Headphones", "price": 2000, "stock": 6}
}

cart = {}

while True:

    print("\n========== SHOPPING CART ==========")
    print("1. View Products")
    print("2. Add to Cart")
    print("3. View Cart")
    print("4. Checkout")
    print("5. Exit")

    choice = input("\nEnter your choice (1-5): ")

    if choice == "1":

        print("\n========== AVAILABLE PRODUCTS ==========")
        print("{:<12} {:<15} {:<15} {:<10}".format(
            "Product ID", "Name", "Price (Rs.)", "Stock"
        ))

        print("-" * 55)

        for product_id, details in products.items():
            print("{:<12} {:<15} {:<15} {:<10}".format(
                product_id,
                details["name"],
                details["price"],
                details["stock"]
            ))

    elif choice == "2":

        product_id = input("\nEnter Product ID: ").upper()

        if product_id in products:

            quantity = int(input("Enter Quantity: "))

            if quantity <= 0:
                print("Please enter a valid quantity.")

            else:
                existing_quantity = 0

                if product_id in cart:
                    existing_quantity = cart[product_id]["quantity"]

                total_quantity = existing_quantity + quantity

                if total_quantity <= products[product_id]["stock"]:

                    if product_id in cart:
                        cart[product_id]["quantity"] += quantity

                    else:
                        cart[product_id] = {
                            "name": products[product_id]["name"],
                            "price": products[product_id]["price"],
                            "quantity": quantity
                        }

                    print("Product added to cart successfully!")

                else:
                    print("Sorry! Not enough stock available.")
                    print("Available Stock:", products[product_id]["stock"])
                    print("Already in Cart:", existing_quantity)

        else:
            print("Invalid Product ID. Please try again.")

    elif choice == "3":

        if len(cart) == 0:
            print("\nYour shopping cart is empty.")

        else:
            print("\n========== YOUR SHOPPING CART ==========")

            total = 0

            print("{:<15} {:<15} {:<10} {:<15}".format(
                "Product", "Price", "Quantity", "Subtotal"
            ))

            print("-" * 55)

            for product_id, details in cart.items():

                subtotal = details["price"] * details["quantity"]
                total += subtotal

                print("{:<15} {:<15} {:<10} {:<15}".format(
                    details["name"],
                    details["price"],
                    details["quantity"],
                    subtotal
                ))

            print("-" * 55)
            print("Cart Total: Rs.", total)

    elif choice == "4":

        if len(cart) == 0:
            print("\nYour cart is empty. Please add products first.")

        else:
            total = 0

            print("\n========== FINAL BILL ==========")

            for product_id, details in cart.items():

                subtotal = details["price"] * details["quantity"]
                total += subtotal

                print(details["name"], "- Quantity:", details["quantity"],
                      "- Subtotal: Rs.", subtotal)

            print("\nTotal Amount: Rs.", total)

            if total >= 5000:
                discount = total * 0.10
            else:
                discount = 0

            final_bill = total - discount

            print("Discount: Rs.", discount)
            print("Final Bill: Rs.", final_bill)

            for product_id, details in cart.items():

                products[product_id]["stock"] -= details["quantity"]

            cart.clear()

            print("\nCheckout Successful!")
            print("Thank you for shopping with us.")

    elif choice == "5":

        print("\nThank you for visiting our store!")
        print("Goodbye!")
        break

    else:
        print("\nInvalid choice! Please select a number from 1 to 5.")
```
