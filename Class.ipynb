```python
class Employee:
    def __init__(self,name,age,salary):
        self.name = name
        self.age = age
        self.salary = salary

    def work(self):
        print(f"{self.name} is working")
```

```python
employee = Employee("Shahbaz", 24, 20000000)
```

```python
employee.name
```

```python
employee.work()
```

```python
class Manager(Employee):
    def __init__(self,name,age,salary, team_size):
        self.name = name
        self.age = age
        self.salary = salary
        self.team_size = team_size

    def getTeamSize(self):
        print(f"Total Team Size: {team_size}")
```

```python
manager = Manager("Nasir", 35, 2000000, 35)
```

```python
manager.work()
```

```python
import time

def calculate_time(func):

    def wrapper():
        start = time.time()

        func()

        end = time.time()

        print(f"Execution time: {end - start:.4f} seconds")

    return wrapper


@calculate_time
def my_task():
    time.sleep(2)
    print("Task completed!")


my_task()
```
